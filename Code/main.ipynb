```python
from sklearn.datasets import make_classification, make_regression, make_blobs
import pandas as pd
import numpy as np
from faker import Faker
```

## Generación de Datos Sintéticos con sklearn

```python
# Generar 1000 muestras con 4 variables (2 útiles para el modelo)
X, y = make_classification(
    n_samples=1000, 
    n_features=4, 
    n_informative=2, 
    n_redundant=0, 
    n_classes=2, 
    random_state=42
)

# Convertir a un DataFrame de Pandas para mejor visualización
df = pd.DataFrame(X, columns=['Variable_1', 'Variable_2', 'Variable_3', 'Variable_4'])
df['Objetivo'] = y
print(df.head())
```

## Generación de Datos Sintéticos con pandas y numpy

```python
# Fijar semilla para reproducibilidad
np.random.seed(42)

# Generar 500 registros
n_registros = 500

datos = {
    'Edad': np.random.randint(18, 65, n_registros),
    'Ingresos': np.random.normal(50000, 15000, n_registros), # Distribución normal
    'Categoria': np.random.choice(['A', 'B', 'C'], n_registros, p=[0.5, 0.3, 0.2]),
    # Variable objetivo basada en una regla lógica con ruido introducido
    'Compra': np.where(np.random.rand(n_registros) > 0.3, 1, 0) 
}

df_custom = pd.DataFrame(datos)
print(df_custom.describe())
```

## Generación de Datos Sintéticos con Faker

```python
fake = Faker('es_ES') # Configurado para español de España
n_registros = 5

datos_usuarios = []
for _ in range(n_registros):
    datos_usuarios.append({
        'Nombre': fake.name(),
        'Ciudad': fake.city(),
        'Trabajo': fake.job(),
        'Email': fake.email()
    })

df_faker = pd.DataFrame(datos_usuarios)
print(df_faker)
```
